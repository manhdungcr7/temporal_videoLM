**Tài khoản Kaggle #1.** GPU **T4 ×2**, bật **Internet** (Session options), rồi **Save Version → Save & Run All**. Thời gian ≈ 4–5 giờ. Khi xong, tải **`j1_event_perception.zip`** ở cell cuối và gửi cho Claude.

Notebook tự chạy thử 4 mẫu trước; nếu có lỗi thật (không tính clip quá dài), nó dừng ngay để không phí GPU. Job tự dừng an toàn trước hạn 12 giờ của Kaggle và giữ những gì đã chạy xong.

### Vì sao chạy
Phân tích hidden state có sẵn (rootcause, n=35–51) cho thấy: vị trí lời nói (nửa đầu/nửa sau) đọc ra được 94–100%, nhưng nhãn TF chỉ 71% (= đúng bằng luật vị trí) và nhãn TA ở mức may rủi. Tức là bên trong mô hình **không có thông tin sự kiện nằm ở nửa nào**. Còn hai khả năng: (P) mô hình không nhìn thấy hành động, hoặc (B) nhìn thấy nhưng không gắn với thời điểm lời nói.

### Đo gì
Với 50 câu mỗi nhãn của TA và TF: câu trả lời gốc + hidden state (token câu hỏi, video từng nửa, audio trong/ngoài lời nói); mô hình tự nghe "người nói nói hành động gì"; rồi hỏi riêng từng nửa video "có thấy hành động đó không". Whisper chỉ dùng để chấm (cận trên của việc nghe).

### Đọc kết quả thế nào (tiêu chí định trước)
**Tri giác ≥ 80% trong khi base ≈ 50%** → nguyên nhân là gắn kết thời gian (B), phương pháp phải sửa ở bước gắn kết. **Tri giác ≈ base** → nguyên nhân là tri giác (P), cần hướng khác.

Không có nhãn nào được đưa vào mô hình; nhãn chỉ dùng để chấm điểm.
